```python
# Cell 2 — Fix Headers and Clean Events Table

import pandas as pd
from pathlib import Path

EVENTS_FILE = Path(
    r"C:\Projects\Infer RozviDrought\data\events\Events.xlsx"
)

# Reload while skipping the title row
events_df = pd.read_excel(EVENTS_FILE, header=1)

# Clean column names
events_df.columns = [str(col).strip() for col in events_df.columns]

print("Cleaned Shape:", events_df.shape)

print("\nCleaned Columns:")
for col in events_df.columns:
    print("-", col)

# Preview dataset
events_df.head()
```
